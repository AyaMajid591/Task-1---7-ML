```python
import numpy as np

from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    balanced_accuracy_score
)
```

```python
TP = 50
FN = 40
FP = 32
TN = 378

print("TP =", TP)
print("FN =", FN)
print("FP =", FP)
print("TN =", TN)
```

```python
manual_matrix = np.array([
    [TP, FN],
    [FP, TN]
])

print(manual_matrix)
```

```python
accuracy = (TP + TN) / (TP + TN + FP + FN)
precision = TP / (TP + FP)
recall = TP / (TP + FN)
specificity = TN / (TN + FP)
f1 = 2 * precision * recall / (precision + recall)

print(f"Accuracy:    {accuracy:.3f}")
print(f"Precision:   {precision:.3f}")
print(f"Recall:      {recall:.3f}")
print(f"Specificity: {specificity:.3f}")
print(f"F1-score:    {f1:.3f}")
```

```python
print(f"Accuracy = ({TP} + {TN}) / ({TP} + {TN} + {FP} + {FN}) = {accuracy:.3f}")
print(f"Precision = {TP} / ({TP} + {FP}) = {precision:.3f}")
print(f"Recall = {TP} / ({TP} + {FN}) = {recall:.3f}")
print(f"Specificity = {TN} / ({TN} + {FP}) = {specificity:.3f}")
print(f"F1 = 2 * {precision:.3f} * {recall:.3f} / ({precision:.3f} + {recall:.3f}) = {f1:.3f}")
```

```python
total = TP + FN + FP + TN

actual_no = FP + TN
always_no_accuracy = actual_no / total

print(f"Always-No Accuracy: {always_no_accuracy:.3f}")
print(f"Our Model Accuracy:  {accuracy:.3f}")
```

The model accuracy is 0.856, while a model that always predicts that the customer will stay has an accuracy of 0.820.

Therefore, accuracy alone can be misleading because most customers stay. Precision, recall, and F1-score should also be considered.

A false negative is more costly in this scenario because the company fails to identify a customer who is actually going to leave.

A false positive may only lead to an unnecessary retention offer, while a false negative may result in losing the customer.

Therefore, recall should be maximized.

```python
TP_new = TP + 25
FN_new = FN - 25

FP_new = FP + 85
TN_new = TN - 85

print("New TP =", TP_new)
print("New FN =", FN_new)
print("New FP =", FP_new)
print("New TN =", TN_new)
```

```python
precision_new = TP_new / (TP_new + FP_new)
recall_new = TP_new / (TP_new + FN_new)
f1_new = 2 * precision_new * recall_new / (precision_new + recall_new)

print(f"New Precision: {precision_new:.3f}")
print(f"New Recall:    {recall_new:.3f}")
print(f"New F1-score:  {f1_new:.3f}")
```

```python
print("Original vs New")
print(f"Precision: {precision:.3f} -> {precision_new:.3f}")
print(f"Recall:    {recall:.3f} -> {recall_new:.3f}")
print(f"F1-score:  {f1:.3f} -> {f1_new:.3f}")
```

```python
y_true = np.array(
    [1] * TP +
    [1] * FN +
    [0] * FP +
    [0] * TN
)

y_pred = np.array(
    [1] * TP +
    [0] * FN +
    [1] * FP +
    [0] * TN
)

print("Number of samples:", len(y_true))
```

```python
cm = confusion_matrix(y_true, y_pred)

print(cm)
```

```python
print(classification_report(y_true, y_pred, digits=3))
```

```python
sk_accuracy = accuracy_score(y_true, y_pred)
sk_precision = precision_score(y_true, y_pred)
sk_recall = recall_score(y_true, y_pred)
sk_f1 = f1_score(y_true, y_pred)

print(f"Accuracy:  {sk_accuracy:.3f}")
print(f"Precision: {sk_precision:.3f}")
print(f"Recall:    {sk_recall:.3f}")
print(f"F1-score:  {sk_f1:.3f}")
```

```python
balanced_accuracy_manual = (recall + specificity) / 2

print(f"Balanced Accuracy (manual): {balanced_accuracy_manual:.3f}")
```

```python
mcc_manual = (
    (TP * TN) - (FP * FN)
) / np.sqrt(
    (TP + FP) *
    (TP + FN) *
    (TN + FP) *
    (TN + FN)
)

print(f"MCC (manual): {mcc_manual:.3f}")
```

```python
mcc_sklearn = matthews_corrcoef(y_true, y_pred)
balanced_accuracy_sklearn = balanced_accuracy_score(y_true, y_pred)

print(f"MCC (sklearn): {mcc_sklearn:.3f}")
print(f"Balanced Accuracy (sklearn): {balanced_accuracy_sklearn:.3f}")
```

```python
print("Manual vs sklearn")
print(f"MCC:               {mcc_manual:.3f} vs {mcc_sklearn:.3f}")
print(f"Balanced Accuracy: {balanced_accuracy_manual:.3f} vs {balanced_accuracy_sklearn:.3f}")
```

MCC is the fairest overall metric for this model. Accuracy is relatively high at 0.856, but it can be misleading because most customers stay. F1-score focuses on the positive class and is useful for churn detection, but it does not consider true negatives. MCC considers TP, TN, FP, and FN at the same time, so it gives a more balanced evaluation of the model.
