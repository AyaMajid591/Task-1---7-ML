```python

import numpy as np

from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
```

Task 7

```python
TP = 136
FN = 24
FP = 31
TN = 209

print("TP =", TP)
print("FN =", FN)
print("FP =", FP)
print("TN =", TN)
```

```python
y_true = np.array(
    [1] * TP +
    [1] * FN +
    [0] * FP +
    [0] * TN
)

y_pred = np.array(
    [1] * TP +
    [0] * FN +
    [1] * FP +
    [0] * TN
)

print("Number of samples:", len(y_true))
```

```python
cm = confusion_matrix(y_true, y_pred)

print(cm)
```

```python
print(classification_report(
    y_true,
    y_pred,
    digits=3
))
```

```python
accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred)
recall = recall_score(y_true, y_pred)
f1 = f1_score(y_true, y_pred)

print(f"Accuracy:  {accuracy:.3f}")
print(f"Precision: {precision:.3f}")
print(f"Recall:    {recall:.3f}")
print(f"F1-score:  {f1:.3f}")
```

- TP = 136: 136 objects were actually plastic, and the model correctly predicted them as plastic.
- FN = 24: 24 objects were actually plastic, but the model predicted them as not plastic.
- FP = 31: 31 objects were actually not plastic, but the model predicted them as plastic.
- TN = 209: 209 objects were actually not plastic, and the model correctly predicted them as not plastic.

# Task 3 — Metrics

### Accuracy

Accuracy = (TP + TN) / (TP + TN + FP + FN)

Accuracy = (136 + 209) / (136 + 209 + 31 + 24)

Accuracy = 345 / 400 = 0.863

### Precision

Precision = TP / (TP + FP)

Precision = 136 / (136 + 31)

Precision = 136 / 167 = 0.814

### Recall

Recall = TP / (TP + FN)

Recall = 136 / (136 + 24)

Recall = 136 / 160 = 0.850

### Specificity

Specificity = TN / (TN + FP)

Specificity = 209 / (209 + 31)

Specificity = 209 / 240 = 0.871

### F1-score

F1 = 2 × Precision × Recall / (Precision + Recall)

F1 = 2 × 0.814 × 0.850 / (0.814 + 0.850)

F1 = 0.832

```python
TP = 136
FN = 24
FP = 31
TN = 209

accuracy = (TP + TN) / (TP + TN + FP + FN)
precision = TP / (TP + FP)
recall = TP / (TP + FN)
specificity = TN / (TN + FP)
f1 = 2 * precision * recall / (precision + recall)

print(f"Accuracy:    {accuracy:.3f}")
print(f"Precision:   {precision:.3f}")
print(f"Recall:      {recall:.3f}")
print(f"Specificity: {specificity:.3f}")
print(f"F1-score:    {f1:.3f}")
```

# Task 5 — More Costly Error

A false positive is more costly in this scenario.

A false positive means that a non-plastic object is incorrectly classified as plastic. This can contaminate the plastic recycling stream and reduce the quality of recycled material.

Therefore, precision should be maximized because higher precision means fewer non-plastic objects are incorrectly classified as plastic.

# Task 6 — Lowered Classification Threshold

Original values:

TP = 136  
FN = 24  
FP = 31  
TN = 209

Threshold shift:

16 FN → TP  
17 TN → FP

New values:

TP = 136 + 16 = 152

FN = 24 - 16 = 8

FP = 31 + 17 = 48

TN = 209 - 17 = 192

### New Confusion Matrix

| Actual \ Predicted | Plastic | Not Plastic |
|---|---:|---:|
| Plastic | TP = 152 | FN = 8 |
| Not Plastic | FP = 48 | TN = 192 |

### New Precision

Precision = 152 / (152 + 48)

Precision = 152 / 200 = 0.760

### New Recall

Recall = 152 / (152 + 8)

Recall = 152 / 160 = 0.950

### New F1-score

F1 = 2 × 0.760 × 0.950 / (0.760 + 0.950)

F1 = 0.844

### Conclusion

After lowering the threshold, recall increased from 0.850 to 0.950 and F1 increased from 0.832 to 0.844.

However, precision decreased from 0.814 to 0.760 because the number of false positives increased.

Since false positives are considered more costly in this scenario, the lower threshold is not necessarily better for the business objective.
